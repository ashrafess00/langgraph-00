# Looping Graph

```python
from typing import TypedDict, List, Dict
from langgraph.graph import StateGraph, END
import random
```

```python
class AgentState(TypedDict):
    name: str
    number: List[int]
    counter: int
```

```python
def greeting_node(state: AgentState) -> AgentState:
    """Greets the user and initializes the number list."""
    state['name'] = f"Hi dear {state['name']}"
    state['counter'] = 0
    return state

def random_number_node(state: AgentState) -> AgentState:
    """Generates a random number and appends it to the number list."""
    rand_num = random.randint(1, 100)
    state['number'].append(rand_num)
    state['counter'] += 1
    return state

def should_continue(state: AgentState) -> str:
    """Decides whether to continue generating numbers based on the counter."""
    if state['counter'] < 5:
        print("Entering loop again...", state["counter"])
        return "true"
    else:
        print("Exiting loop...", state["counter"])
        return "false"
```

```python
graph = StateGraph(AgentState)

graph.add_node('greeting_node', greeting_node)
graph.add_node('random_number_node', random_number_node)

graph.add_edge('greeting_node', 'random_number_node')

graph.add_conditional_edges(
    'random_number_node',
    should_continue,
    {
        "true": 'random_number_node',
        "false": END
    }
)

graph.set_entry_point('greeting_node')

app = graph.compile()
```

```python
app.invoke({
    'name': 'Alice',
    'number': [],
    'counter': 0})
```
