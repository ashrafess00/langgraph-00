```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    age: str
    final: str
```

```python
def first_node(state: AgentState) -> AgentState:
    """ First node to get the name of the user. """
    state["final"] = f"Hi {state['name']}, nice to meet you!"
    return state

def second_node(state: AgentState) -> AgentState:
    """ Second node to get the age of the user. """
    state["final"] = state["final"] + f" Wow, {state['age']} is a great age to be!"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("first_node", first_node)
graph.add_node("second_node", second_node)

graph.set_entry_point("first_node")
graph.add_edge("first_node", "second_node")
graph.set_finish_point("second_node")

app = graph.compile()
```

```python
result = app.invoke({"name": "Alice", "age": "30"})
print(result["final"])  # Output: Hi Alice, nice to meet you! Wow, 30 is a great age to be!
```
